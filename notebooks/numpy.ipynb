**This notebook contains the implementation of Numpy**

```python
import numpy as np
```

```python
print(np.__version__)
```

```python
arr=np.array([1,2,3,4,5]) #This is vectorised and is very different form standard python list.
print(arr*2)
```

```python
zeroD=np.array([1]) #single point
oneD=np.array([1,2,3]) #one dimensional array or vector
twoD=np.array([[1,2],[1,2]]) #two dimensional array or vector
threeD=np.array([[[1,2],[1,2],[1,2]],[[1,2],[1,2],[1,2]]]) #3D vector

print(threeD.ndim)
print(threeD.shape) #(layer,row,column)
print(threeD)
```

```python
print(threeD[0,0,1]) #numpy allows you to access elemets directly inside one square bracket.
print(twoD[1,0])
```

*Slicing*

```python
print(threeD[0,0:2,0:2]) #array[depth_start:depth_stop, row_start:row_stop, col_start:col_stop]
print(twoD[0:2,:]) #last element is exclusive.
```

```python
a = np.arange(24).reshape(2, 3, 4)
print(a)
```

```python
print(a[0,0:2,1:3])
```

**Arithmetic**

```python
x=np.array([[4,5,6],[1,2,4],[4,2,5]])
y=np.array([[4,3,7],[9,2,4],[0,1,6]])

print(x + 10) #scalar addition , added to each element
print(x*10) #scalar multiplication , each element multiplied
print(x*y) #matrix multiplication
print(x>5) #each element compared
x[x>2]=0 #selecting all the elements that are greater than 2 and setting them to zero.
print(x)
```

## Aggregate Functions

*Definition*: Functions that summarize data into a single value, such as finding the sum, mean, or maximum. You can specify an axis to aggregate across columns (axis=0) or rows (axis=1).

While there are over 20 functions, you can conceptually remember them as 5 categories: Averages, Sums, Extrema, Logic, and NaN-Safe versions.

In real-world data science, you often have missing data represented as NaN (Not a Number). If you try to use np.sum() on an array containing a NaN, the result will just be NaN. Do use NaN safe version in such data.

*1. Sums & Products (Basic Arithmetic)*
·	np.sum(): Adds all elements together.

·	np.prod(): Multiplies all elements together (returns the product).

·	np.cumsum(): Returns an array of the cumulative sum (a running total at each index).

·	np.cumprod(): Returns an array of the cumulative product (running multiplication).

*2. Averages & Spread (Statistics)*

·	np.mean(): Calculates the arithmetic average.

·	np.median(): Finds the exact middle value (halving the sorted data).

·	np.std(): Computes the Standard Deviation (measures how spread out the data is from the mean).

·	np.var(): Computes the Variance (the average of the squared deviations).

·	np.ptp(): "Peak-to-Peak" — calculates the total range of the data (Maximum−Minimum).

*3. Extrema (Min/Max and Their Positions)*

·	np.min() (or np.amin()): Finds the smallest value in the array.

·	np.max() (or np.amax()): Finds the largest value in the array.

·	np.argmin(): Returns the index position of the smallest value, rather than the value itself.

·	np.argmax(): Returns the index position of the largest value.

*4. Logic & Conditions (Boolean Aggregates)*

·	np.all(): Checks if every single element evaluates to True (works like an AND gate).

·	np.any(): Checks if at least one element evaluates to True (works like an OR gate).

*5. NaN-Safe Aggregates (Ignoring Missing Data)*

·	np.nansum(): Sums elements, treating NaN as zero.

·	np.nanprod(): Multiplies elements, treating NaN as one.

·	np.nanmean(): Computes the average of non-missing values.

·	np.nanmedian(): Finds the middle value of non-missing data.

·	np.nanstd(): Computes standard deviation, ignoring NaNs.

·	np.nanvar(): Computes variance, ignoring NaNs.

·	np.nanmin(): Finds the minimum valid value.

·	np.nanmax(): Finds the maximum valid value.

·	np.nanargmin(): Finds the index of the lowest valid value.

·	np.nanargmax(): Finds the index of the highest valid value.




```python
print(np.sum(a1))
print(np.mean(a1))
print(np.var(a1))
print(np.min(a1))
print(np.max(a1))
print(np.cumsum(a1))
print(np.cumprod(a1))
print(np.median(a1))
print(np.ptp(a1))
print(np.argmax(a1))
```

**Filtering Functions**

* **`arr[condition]` (Boolean Indexing):** Directly extracts elements that meet the condition. 
  * *Behavior:* Drops all `False` values and **flattens** the output into a 1D array.
* **`np.where(condition, if_true, if_false)` (If-Else Replacer):** Replaces array values based on a condition. 
  * *Behavior:* Elements meeting the condition get the `if_true` value; all others get the `if_false` value. **Preserves the original shape.**
* **`np.where(condition)` (Index Locator):** Finds the exact positions of elements that match the condition. 
  * *Behavior:* Returns a **tuple of coordinate arrays** mapping the indices for each dimension.


```python
print(np.array(a1[a1>2]))
print(np.array(np.where(a1>2,a1,-1)))
```

# Quick Summary: Stacking and Concatenation

* **`np.concatenate((arr1, arr2), axis)`:** Glues arrays together along a pre-existing dimension.
  * *Behavior:* Combines arrays end-to-end along the specified `axis` (defaults to axis 0).
* **`np.vstack((arr1, arr2))` (Vertical Stack):** Stacks arrays row-wise on top of each other.
  * *Behavior:* Increases the number of rows; converts 1D arrays into 2D rows automatically.
* **`np.hstack((arr1, arr2))` (Horizontal Stack):** Stacks arrays column-wise side-by-side.
  * *Behavior:* Increases the number of columns; links 1D arrays end-to-end.


```python
print(np.vstack((mA,mB)))
print(np.hstack((mA,mB)))
print(np.concatenate((mA,mB),axis=0)) #same as vstack
print(np.concatenate((mA,mB),axis=1)) #same as hstack
#remember the bracket placement

 
```

# Shape Manipulation Summary & Examples

* **`np.reshape(arr, new_shape)`:** Changes the dimensions of an array.
  * *Behavior:* Re-arranges the data grid. Using `-1` tells NumPy to automatically calculate that specific dimension's size.
* **`.flatten()` vs `.ravel()`:** Collapses a multi-dimensional array into a single 1D vector.
  * *Behavior:* `.flatten()` creates a completely new **copy** of the data; `.ravel()` creates a memory-efficient **view** (modifying a ravelled array changes the original).
* **`np.newaxis` or `np.expand_dims(arr, axis)`:** Adds a "dummy" dimension of size 1.
  * *Behavior:* Increases the total number of dimensions; crucial for matching target shapes in Deep Learning models.
  
Think of np.newaxis and np.expand_dims as adding an extra set of outer brackets [] around your data without changing any of the actual numbers.

The axis parameter in np.expand_dims(arr, axis) specifies the index position where a new length-1 dimension is inserted. For a 2D array with shape (3, 3):   axis=0 $\rightarrow$ (1, 3, 3): Inserts a new dimension at the front, wrapping the entire matrix into a single container.   axis=1 $\rightarrow$ (3, 1, 3): Inserts a new dimension in the middle, wrapping each individual row.   axis=2 $\rightarrow$ (3, 3, 1): Inserts a new dimension at the end, wrapping every single number.   

```python
print(np.reshape(mA,(3,-1)))
print(mA.flatten())
```

```python
print(np.expand_dims(mA,0))
print(np.expand_dims(mA,1))
print(np.expand_dims(mA,2))
```

# Sorting

* **`np.sort(arr, axis)`:** Sorts the elements of an array along a specified direction.
  * *Behavior:* Returns a sorted **copy** of the array. By default (`axis=-1`), it sorts along the last axis (columns).
* **`np.argsort(arr, axis)`:** Finds the indices that would sort the array.
  * *Behavior:* Instead of returning the sorted numbers, it returns the **index positions** of where those numbers came from. Extremely useful for sorting a second dataset based on the first.

# Random Numbers & Shuffling

*Definition*: Using a Random Number Generator (RNG) to initialize model weights, create synthetic data, or shuffle datasets. A **seed** ensures the "random" results are reproducible across runs.

Random numbers in NumPy aren't truly random — they're **pseudo-random**. The algorithm takes a starting number (the **seed**), runs it through a deterministic mathematical formula, and produces a sequence that looks random.

* **`rng = np.random.default_rng(seed=42)`:** Creates a modern Generator instance with a fixed seed (reproducible).
* **`rng = np.random.default_rng()`:** Creates a Generator without a seed (unpredictable, pulls entropy from the OS).

*1. Core Distributions for ML*
* **`rng.standard_normal(size)`:** Draws from a standard Gaussian distribution ($\mu = 0, \sigma = 1$). Crucial for initializing neural network weights.
* **`rng.normal(loc, scale, size)`:** Draws from a custom Gaussian distribution (`loc` = mean, `scale` = standard deviation). Used for adding controlled noise.
* **`rng.uniform(low, high, size)`:** Draws uniformly distributed floats between `low` (inclusive) and `high` (exclusive). Used for random hyperparameter search.
* **`rng.integers(low, high, size)`:** Draws random integers from `low` (inclusive) to `high` (exclusive). Useful for generating mock class labels.
* **`rng.choice(arr, size, replace)`:** Randomly samples elements from an existing array (`replace=False` for unique mini-batches, `replace=True` for bootstrapping).

*2. Shuffling Safely in ML (`shuffle` vs `permutation`)*
* **`rng.shuffle(arr)`:** Shuffles the array **in-place** (modifies the original array and returns `None`). **Warning:** Calling this separately on features `X` and labels `y` will misalign your dataset!
* **`rng.permutation(n)`:** Returns a **new shuffled array of indices** from `0` to `n-1` without modifying the original data. Pass these indices into both `X` and `y` using Fancy Indexing to keep features and labels aligned.

```python
rng=np.random.default_rng(seed=1)
print(rng.integers(low=1,high=100,size=3)) #gives a list
print(rng.integers(low=1,high=100,size=(3,3))) #gives a 2D 3x3 array


```

```python
np.random.seed(seed=1)
print(np.random.uniform(low=-1,high=1,size=(3,3,3))) #floating point number random generator
```

```python
rng.shuffle(mA) #This method shuffles the array directly in place .
print(mA)
```

```python
print(rng.choice(mA,size=(3,3)))
```

## Array Creation & Initialization

**Definition**: Functions that generate pre-filled arrays of a specific shape or numerical sequence without typing out lists manually. In Machine Learning, these are essential for initializing bias vectors, creating masks, one-hot encoding labels, and generating coordinate grids to plot decision boundaries.

While there are many creation functions, you can conceptually remember them in 3 categories: Constant-Filled Arrays, Shape-Matching (`_like`) Arrays, and Sequence Generators.

**1. Constant-Filled & Identity Arrays**
·	np.zeros(shape): Creates an array filled entirely with `0.0` (defaults to float64). Commonly used to initialize bias vectors in neural networks.

·	**np.ones(shape)**: Creates an array filled entirely with `1.0`. Frequently used to prepend a column of ones to a feature matrix $X$ (the "bias trick" in Linear Regression).

·	**np.full(shape, fill_value)**: Creates an array of a given shape filled entirely with any custom constant value you specify.

·	**np.empty(shape)**: Allocates an uninitialized array of the given shape (contains whatever random garbage values were already in memory). Slightly faster than `zeros()` when you plan to immediately overwrite every element.

·	**np.eye(N) (or np.identity(N))**: Creates an $N \times N$ Identity Matrix (`1.0` on the main diagonal and `0.0` everywhere else). Crucial for linear algebra and instant One-Hot Encoding of target classes.

·	**np.diag(v)**: Extracts the diagonal of a 2D matrix into a 1D array, OR converts a 1D array `v` into a 2D matrix with `v` along the diagonal.

*2. Shape-Matching Arrays (The `_like` Family)*
·	**np.zeros_like(arr)**: Creates a new array of `0`s that automatically matches the exact `shape` and `dtype` of an existing array `arr`.

·	**np.ones_like(arr)**: Creates a new array of `1`s matching the exact `shape` and `dtype` of `arr`.

·	**np.full_like(arr, fill_value)**: Creates a new array filled with `fill_value`, matching the exact `shape` and `dtype` of `arr`.

*3. Sequence & Grid Generators*
·	**np.arange(start, stop, step)**: Generates evenly spaced values within a half-open interval `[start, stop)` (where `stop` is exclusive) using a specific **step size**.

·	**np.linspace(start, stop, num)**: Generates `num` evenly spaced values over a closed interval `[start, stop]` (**`stop` is inclusive** by default!). Ideal for generating smooth axes of 100+ points when plotting ML decision boundaries or learning curves.

```python
print(np.zeros((3,3)))
print(np.ones((3,3)))
print(np.full((3,3),-1))
print(np.eye(3))
print(np.diag(mA))
print(np.zeros_like(mA))
```

## Axis Aggregations (keepdims), Fancy Indexing & Clipping

*Definition*: Specialized indexing, boundary control, and dimensional preservation tools needed for training neural networks and evaluating loss functions.

*1. Multi-Dimensional Aggregations & `keepdims`*
·	axis=0 vs axis=1: In a 2D data matrix $(N \times D)$ where rows are samples and columns are features:
  - `axis=0` aggregates down columns $\rightarrow$ returns 1 value per feature (used for feature scaling).
  - `axis=1` aggregates across rows $\rightarrow$ returns 1 value per sample (used for row normalization/softmax).

·	keepdims=True: Retains the reduced axes as dimensions with size 1 instead of collapsing them into a 1D vector. For instance, aggregating a $(2, 3)$ matrix along `axis=1` produces $(2, 1)$ with `keepdims=True` versus $(2,)$ without it. This ensures clean broadcasting back against the original matrix.

*2. Fancy Indexing & Categorical Encoding*
·	arr[[i, j, k]]: Passing an integer array or list of coordinates extracts non-contiguous rows or elements. Unlike slicing, fancy indexing always produces a independent **copy**.

·	np.eye(C)[labels]: Indexing a $C \times C$ identity matrix with an integer vector of class labels instantly generates an $(N, C)$ **One-Hot Encoded** matrix.

*3. Numerical Stability*
·	np.clip(arr, a_min, a_max): Forces all values below `a_min` to `a_min` and above `a_max` to `a_max`. Essential for avoiding division by zero or taking the log of zero in loss functions like Binary Cross-Entropy.

```python
# 1. Axis Aggregations with keepdims=True
X_features = np.array([[10.0, 20.0, 30.0],
                       [40.0, 50.0, 60.0]]) # Shape: (2, 3)

# Row-wise sum with keepdims keeps shape as (2, 1), enabling clean division
row_sums = np.sum(X_features, axis=1, keepdims=True)
normalized_X = X_features / row_sums
print("Normalized Rows (shape:", normalized_X.shape, "):\n", normalized_X)

# 2. Fancy Indexing & One-Hot Encoding
data_matrix = np.array([[100, 101], [200, 201], [300, 301], [400, 401]])
row_select = np.array([0, 3, 1])
print("Selected Rows (Fancy Indexing):\n", data_matrix[row_select])

target_labels = np.array([0, 2, 1, 0])
one_hot = np.eye(3)[target_labels] # 3 classes: 0, 1, 2
print("One-Hot Encoded Matrix:\n", one_hot)

# 3. Value Clipping for Loss Stability
predicted_probs = np.array([0.0, 0.45, 0.99, 1.0])
safe_probs = np.clip(predicted_probs, 1e-15, 1.0 - 1e-15)
print("Clipped Probabilities:", safe_probs)
```

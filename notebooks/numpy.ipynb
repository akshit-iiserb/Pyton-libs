## Aggregate Functions

*Definition*: Functions that summarize data into a single value, such as finding the sum, mean, or maximum. You can specify an axis to aggregate across columns (axis=0) or rows (axis=1).

While there are over 20 functions, you can conceptually remember them as 5 categories: Averages, Sums, Extrema, Logic, and NaN-Safe versions.

In real-world data science, you often have missing data represented as NaN (Not a Number). If you try to use np.sum() on an array containing a NaN, the result will just be NaN. Do use NaN safe version in such data.

*1. Sums & Products (Basic Arithmetic)*
·	np.sum(): Adds all elements together.

·	np.prod(): Multiplies all elements together (returns the product).

·	np.cumsum(): Returns an array of the cumulative sum (a running total at each index).

·	np.cumprod(): Returns an array of the cumulative product (running multiplication).

*2. Averages & Spread (Statistics)*

·	np.mean(): Calculates the arithmetic average.

·	np.median(): Finds the exact middle value (halving the sorted data).

·	np.std(): Computes the Standard Deviation (measures how spread out the data is from the mean).

·	np.var(): Computes the Variance (the average of the squared deviations).

·	np.ptp(): "Peak-to-Peak" — calculates the total range of the data (Maximum−Minimum).

*3. Extrema (Min/Max and Their Positions)*

·	np.min() (or np.amin()): Finds the smallest value in the array.

·	np.max() (or np.amax()): Finds the largest value in the array.

·	np.argmin(): Returns the index position of the smallest value, rather than the value itself.

·	np.argmax(): Returns the index position of the largest value.

*4. Logic & Conditions (Boolean Aggregates)*

·	np.all(): Checks if every single element evaluates to True (works like an AND gate).

·	np.any(): Checks if at least one element evaluates to True (works like an OR gate).

*5. NaN-Safe Aggregates (Ignoring Missing Data)*

·	np.nansum(): Sums elements, treating NaN as zero.

·	np.nanprod(): Multiplies elements, treating NaN as one.

·	np.nanmean(): Computes the average of non-missing values.

·	np.nanmedian(): Finds the middle value of non-missing data.

·	np.nanstd(): Computes standard deviation, ignoring NaNs.

·	np.nanvar(): Computes variance, ignoring NaNs.

·	np.nanmin(): Finds the minimum valid value.

·	np.nanmax(): Finds the maximum valid value.

·	np.nanargmin(): Finds the index of the lowest valid value.

·	np.nanargmax(): Finds the index of the highest valid value.




```python
print(np.sum(a1))
print(np.mean(a1))
print(np.var(a1))
print(np.min(a1))
print(np.max(a1))
print(np.cumsum(a1))
print(np.cumprod(a1))
print(np.median(a1))
print(np.ptp(a1))
print(np.argmax(a1))
```

**Filtering Functions**

* **`arr[condition]` (Boolean Indexing):** Directly extracts elements that meet the condition. 
  * *Behavior:* Drops all `False` values and **flattens** the output into a 1D array.
* **`np.where(condition, if_true, if_false)` (If-Else Replacer):** Replaces array values based on a condition. 
  * *Behavior:* Elements meeting the condition get the `if_true` value; all others get the `if_false` value. **Preserves the original shape.**
* **`np.where(condition)` (Index Locator):** Finds the exact positions of elements that match the condition. 
  * *Behavior:* Returns a **tuple of coordinate arrays** mapping the indices for each dimension.


```python
print(np.array(a1[a1>2]))
print(np.array(np.where(a1>2,a1,-1)))
```

# Quick Summary: Stacking and Concatenation

* **`np.concatenate((arr1, arr2), axis)`:** Glues arrays together along a pre-existing dimension.
  * *Behavior:* Combines arrays end-to-end along the specified `axis` (defaults to axis 0).
* **`np.vstack((arr1, arr2))` (Vertical Stack):** Stacks arrays row-wise on top of each other.
  * *Behavior:* Increases the number of rows; converts 1D arrays into 2D rows automatically.
* **`np.hstack((arr1, arr2))` (Horizontal Stack):** Stacks arrays column-wise side-by-side.
  * *Behavior:* Increases the number of columns; links 1D arrays end-to-end.


```python
print(np.vstack((mA,mB)))
print(np.hstack((mA,mB)))
print(np.concatenate((mA,mB),axis=0)) #same as vstack
print(np.concatenate((mA,mB),axis=1)) #same as hstack
#remember the bracket placement

 
```

# Shape Manipulation Summary & Examples

* **`np.reshape(arr, new_shape)`:** Changes the dimensions of an array.
  * *Behavior:* Re-arranges the data grid. Using `-1` tells NumPy to automatically calculate that specific dimension's size.
* **`.flatten()` vs `.ravel()`:** Collapses a multi-dimensional array into a single 1D vector.
  * *Behavior:* `.flatten()` creates a completely new **copy** of the data; `.ravel()` creates a memory-efficient **view** (modifying a ravelled array changes the original).
* **`np.newaxis` or `np.expand_dims(arr, axis)`:** Adds a "dummy" dimension of size 1.
  * *Behavior:* Increases the total number of dimensions; crucial for matching target shapes in Deep Learning models.
  
Think of np.newaxis and np.expand_dims as adding an extra set of outer brackets [] around your data without changing any of the actual numbers.

The axis parameter in np.expand_dims(arr, axis) specifies the index position where a new length-1 dimension is inserted. For a 2D array with shape (3, 3):   axis=0 $\rightarrow$ (1, 3, 3): Inserts a new dimension at the front, wrapping the entire matrix into a single container.   axis=1 $\rightarrow$ (3, 1, 3): Inserts a new dimension in the middle, wrapping each individual row.   axis=2 $\rightarrow$ (3, 3, 1): Inserts a new dimension at the end, wrapping every single number.   

```python
print(np.reshape(mA,(3,-1)))
print(mA.flatten())
```

```python
print(np.expand_dims(mA,0))
print(np.expand_dims(mA,1))
print(np.expand_dims(mA,2))
```

# Sorting

* **`np.sort(arr, axis)`:** Sorts the elements of an array along a specified direction.
  * *Behavior:* Returns a sorted **copy** of the array. By default (`axis=-1`), it sorts along the last axis (columns).
* **`np.argsort(arr, axis)`:** Finds the indices that would sort the array.
  * *Behavior:* Instead of returning the sorted numbers, it returns the **index positions** of where those numbers came from. Extremely useful for sorting a second dataset based on the first.



```python
print(np.sort(mA))
#to short in place use mA.sort().
print(np.argsort(mA))
```
